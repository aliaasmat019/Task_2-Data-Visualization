# Task 2 — Data Visualization

Visual analysis of the Superstore dataset using **Python, Matplotlib, Seaborn, and Power BI**.

The notebook focuses on sales performance across time, region, and category, with visuals designed to support quick comparison and decision-making.

## Analytical Focus

- How do sales change over time?
- Which regions generate the most sales?
- Which product categories contribute most to sales?
- How is sales distributed within each category?
- What does the combined view suggest for business decisions?

## Setup

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
COLOR = "#016171"

df = pd.read_csv("../data/SuperStore_Cleaned.csv")
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.to_period("M").astype(str)

print(f"Rows: {len(df):,}")
print(f"Columns: {df.shape[1]:,}")
```

## KPI Summary

```python
total_sales = df["Sales"].sum()
total_transactions = df["Order ID"].nunique()

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Transactions: {total_transactions:,}")
```

## Sales by Year

```python
year_sales = df.groupby("Year")["Sales"].sum()

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(year_sales.index.astype(str), year_sales.values, color=COLOR)

ax.set_title("Total Sales by Year", fontweight="bold")
ax.set_xlabel("Year")
ax.set_ylabel("Sales")
ax.bar_label(bars, labels=[f"${v:,.0f}" for v in year_sales.values], padding=3, fontsize=9)
plt.tight_layout()
plt.show()
```

## Sales by Region

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values()

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.barh(region_sales.index, region_sales.values, color=COLOR)

ax.set_title("Total Sales by Region", fontweight="bold")
ax.set_xlabel("Sales")
ax.set_ylabel("Region")
ax.bar_label(bars, labels=[f"${v:,.0f}" for v in region_sales.values], padding=3, fontsize=9)
plt.tight_layout()
plt.show()
```

## Sales by Category

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values()

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.barh(category_sales.index, category_sales.values, color=COLOR)

ax.set_title("Total Sales by Category", fontweight="bold")
ax.set_xlabel("Sales")
ax.set_ylabel("Category")
ax.bar_label(bars, labels=[f"${v:,.0f}" for v in category_sales.values], padding=3, fontsize=9)
plt.tight_layout()
plt.show()
```

## Sales Trend

```python
monthly_sales = df.groupby("Month")["Sales"].sum()
monthly_sales.index = pd.to_datetime(monthly_sales.index)

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(monthly_sales.index, monthly_sales.values, color=COLOR, linewidth=2.5, marker="o", markersize=3)

ax.set_title("Monthly Sales Trend", fontweight="bold")
ax.set_xlabel("Order Date")
ax.set_ylabel("Sales")
ax.tick_params(axis="x", rotation=45)
plt.tight_layout()
plt.show()
```

## Sales Distribution by Category

```python
fig, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(data=df, x="Category", y="Sales", color=COLOR, ax=ax)

ax.set_title("Sales Distribution by Category", fontweight="bold")
ax.set_xlabel("Category")
ax.set_ylabel("Sales")
plt.tight_layout()
plt.show()
```

## Data Story

Sales performance is examined from four complementary perspectives: **time, region, category, and distribution**. Together, these views show where sales are concentrated, how performance changes over time, and how variable order-level sales are within categories.

The Power BI dashboard provides the interactive version of the same reporting story, while this notebook provides a reproducible Python implementation.
